# **Analyzing Gender columns**

```python
import pandas as pd
import numpy as np
df = pd.read_csv(r"C:\Users\Rishi kumar\Downloads\BlackFriday.csv")
del df['Product_Category_2']
del df['Product_Category_3']
```

```python
df.head()
```

```python
for column in df.columns:
    print(df[column].nunique(), "\t:",column)
```

```python
#how many males and females 
len(df[df['Gender'] == 'M']),len(df[df['Gender'] == 'F'])
```

```python
#make a data frame of count of gender
data= pd.DataFrame({'Ratio' : [len(df[df['Gender'] == 'M']),len(df[df['Gender'] == 'F'])]}, index = ['Male' , 'Female'])
```

```python
data
```

```python
#piechart of ratio between no of males and females
data.plot.pie(y = 'Ratio', figsize = (6,6), autopct = "%.2f")
```

```python
#one line code using group by and plot a pie chart of ratio between male and female
df.groupby('Gender').size().plot(kind= 'pie',figsize = (6,6), autopct = "%.2f", title = 'Gender Ratio')
```

```python
#one line code using group by and plot a bar chart of ratio between male and female
df.groupby('Gender').size().plot(kind= 'bar',figsize = (6,6),  title = 'Gender Ratio')
```

```python
df.groupby('Gender').size()
```

```python
#total purchase of male and female
df.groupby('Gender')['Purchase'].sum()
```

```python
#a pie chart of total purchase of male and female
df.groupby('Gender')['Purchase'].sum().plot(kind = 'pie' , autopct = "%0.1f")
```

```python
#a pie chart of average purchase of male and female
df.groupby('Gender')['Purchase'].mean().plot(kind = 'pie' , autopct = "%0.1f")
```
