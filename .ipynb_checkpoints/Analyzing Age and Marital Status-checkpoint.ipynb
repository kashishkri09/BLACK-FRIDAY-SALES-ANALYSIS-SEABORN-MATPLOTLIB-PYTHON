```python
import pandas as pd
import numpy as np
df = pd.read_csv(r"C:\Users\Rishi kumar\Downloads\BlackFriday.csv")
del df['Product_Category_2']
del df['Product_Category_3']
df.head()
```

```python
#number of total product in the category
df.groupby('Age').size()
```

```python
df.groupby('Age').size().plot(kind = 'bar', figsize=(12,6), title='Purchase Distribution by Age')
```

```python
#number of total productage puchase in the category
for i in df['Age'].unique():
    print(i, ":", len(df[df['Age']==i]))
```

```python
#unique product purchage in the category and create a list and store in it and convert into dataframe
lst = []
for i in df['Age'].unique():
    lst.append([i, df[df['Age']==i]['Product_ID'].nunique()])
data= pd.DataFrame(lst, columns=['Age','Products'])
data
```

```python
#unique products in the inventory based on the age segment
data.plot.bar(x= 'Age', figsize=(9,6))
```

```python
# bar chart of total amount purchase by age group
df.groupby('Age').sum()['Purchase'].plot(kind = 'bar', figsize=(5,6), title='Amount Spend by Age')
```

```python
# bar chart of mean amount purchase by age group
df.groupby('Age')['Purchase'].mean()
df.groupby('Age')['Purchase'].mean().plot(kind='bar',figsize=(12, 6),title='Average Purchase Amount by Age Group')
```

```python
df.groupby('Age')['Purchase'].mean().plot(kind='pie', autopct = '%0.1F',title='Average Purchase Amount by Age Group' )
```

```python
df.groupby('Marital_Status').size().plot(kind='pie', autopct = '%0.1F' ,title='Marital Status')
```

```python
df.groupby('Gender').size().plot(kind='pie', autopct = '%0.1F', title='Gender')
```
