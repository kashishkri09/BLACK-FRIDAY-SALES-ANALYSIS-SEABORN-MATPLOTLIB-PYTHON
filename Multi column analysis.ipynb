```python
import pandas as pd
import numpy as np
import seaborn as sns
df = pd.read_csv(r"C:\Users\Rishi kumar\Downloads\BlackFriday.csv")
del df['Product_Category_2']
del df['Product_Category_3']
```

```python
df.head()
```

```python
#only one column
df.groupby('Age').size().plot(kind = 'bar', figsize=(6,6), title='Purchase Distribution by Age')
```

```python
#using two or multiplecolumns with seaborn library
#show product purchase between male and female
sns.set(rc={'figure.figsize':(6,6)})
sns.countplot(x='Age', hue='Gender' ,data= df)
```

```python
#show product purchase between gender and age category
sns.set(rc={'figure.figsize':(6,6)})
sns.countplot(x='Gender', hue='Age' ,data= df)
```

```python
#show product purchase between gender and marital status
sns.set(rc={'figure.figsize':(6,6)})
sns.countplot(x='Gender', hue='Marital_Status' ,data= df)
```

```python
sns.set(rc={'figure.figsize':(6,6)})
sns.countplot(x='Marital_Status', hue='Gender' ,data= df)
```

```python
#which city have purchase more product
sns.countplot(x=df['City_Category'])
```

```python
df.groupby('City_Category').size().plot(kind = 'pie', autopct= '%0.1f')
```

```python
#show product purchase between age category and city
sns.countplot(x=df['City_Category'],hue='Age', data=df)
```

```python
sns.countplot(x=df['Age'],hue='City_Category', data=df)
```

```python
#show product purchase between marital status and city
sns.countplot(x=df['Marital_Status'],hue='City_Category', data=df)
```

```python
#m0,m1,f0,f1
```

```python
#show product purchase between city and gender
sns.countplot(x=df['City_Category'],hue='Gender', data=df)
```

```python
#total amount spent by city category
df.groupby('City_Category')['Purchase'].sum().plot(kind= 'pie', autopct= '%0.1f')
#city b spent more amount
```

```python
#average amount spent by city category
df.groupby('City_Category')['Purchase'].mean().plot(kind= 'pie', autopct= '%0.1f')
#city a and b spent almost equal amount that means a purchase higher price product
```

```python
#We combine Age, Marital Status and Gender to see if purchasing patterns vary when users' characteristics are combined.
#By combining both factors with visualizations like pie charts and bar plots, we gained a clearer understanding of purchasing behavior across different age groups and marital statuses.
```
