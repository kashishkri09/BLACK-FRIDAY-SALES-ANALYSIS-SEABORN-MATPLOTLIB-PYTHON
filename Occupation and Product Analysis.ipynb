```python
import pandas as pd
import numpy as np
import seaborn as sns
df = pd.read_csv(r"C:\Users\Rishi kumar\Downloads\BlackFriday.csv")
del df['Product_Category_2']
del df['Product_Category_3']
```

```python
df.head()
```

```python
#bar chart of total count of product purchase in Stay_In_Current_City_Years
sns.countplot(x=df['Stay_In_Current_City_Years'])
```

```python
#pie chart of total count of product purchase in Stay_In_Current_City_Years
df.groupby('Stay_In_Current_City_Years').size().plot(kind='pie', autopct = '%0.1f')
```

```python
#bar chart of total count of product purchase in Stay_In_Current_City_Years between male and female
sns.countplot(x='Stay_In_Current_City_Years', hue = 'Gender', data=df)
```

```python
#bar chart of total count of product purchase in Stay_In_Current_City_Years marital status
sns.countplot(x='Stay_In_Current_City_Years', hue = 'Marital_Status', data=df)
```

```python
#bar chart of total count of product purchase in Stay_In_Current_City_Years between city category
sns.countplot(x='Stay_In_Current_City_Years', hue = 'City_Category', data=df)
```

```python
#bar chart of total count of product purchase in city category and age category
sns.countplot(x='City_Category', hue = 'Age', data=df)
```

```python
#bar chart of total sum of product purchase in Stay_In_Current_City_Years
df.groupby('Stay_In_Current_City_Years')['Purchase'].sum().plot(kind='bar')
```

```python
#bar chart of total mean of product purchase in Stay_In_Current_City_Years
df.groupby('Stay_In_Current_City_Years')['Purchase'].mean().plot(kind='bar')
```

```python
#bar chart of total count of product purchase betwwen occupation and maritial status
sns.countplot(x='Occupation', hue= 'Marital_Status', data=df)
```

```python
#bar chart of total count of product purchase betwwen occupation and gender
sns.countplot(x='Occupation', hue= 'Gender', data=df)
```
